# `*args` vs `**kwargs`

```python
# *args — collects positional args into a tuple
def func(*args):
    print(args)

func(1, 2, 3)

nums = [1, 2, 3]
func(*nums)
```

```python
# **kwargs — collects keyword args into a dict
def func(**kwargs):
    print(kwargs)

func(a=1, b=2)

data = {'a': 1, 'b': 2}
func(**data)
```

| | `*args` | `**kwargs` |
|---|---|---|
| input | positional `1, 2, 3` | keyword `a=1, b=2` |
| type | `tuple` | `dict` |
| output | `(1, 2, 3)` | `{'a': 1, 'b': 2}` |
