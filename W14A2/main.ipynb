Develop a simple function such as add(a, b) and test it. Then, build a decorator called log_decorator that prints the function name, input arguments, and result. Apply the decorator using @log_decorator and run the function again to observe the logging output. ## 1) `add(a, b)` — test without decorator

```python
def add(a, b):
    return a + b


print("Test add without decorator:", add(2, 3))
```

## 2) `log_decorator` — print function name, inputs, result

```python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        print("--- log ---")
        print("function:", func.__name__)
        print("arguments:", args, kwargs)
        print("result:", result)
        print("-----------")
        return result

    return wrapper
```

## 3) `add` again with `@log_decorator`

```python
@log_decorator
def add(a, b):
    return a + b


print("Test add with @log_decorator:")
add(2, 3)
```

## 4) `multiply` — same decorator

```python
@log_decorator
def multiply(a, b):
    return a * b


print("Test multiply with @log_decorator:")
multiply(4, 5)
```
